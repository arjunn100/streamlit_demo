```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Name": ["A","B","C","D","E"],
    "Maths": [78,67,20,95,62],
    "Science": [35,65,89,65,32],
    "English": [49,69,32,56,30] 
}
df = pd.DataFrame(data)
print(df)
```

```python
marks = df[["Maths", "Science", "English"]].values
average_marks = np.mean(marks, axis =1)
df["Average"] = average_marks
print(df)
```

```python
#assign grade
def grade(avg):
    if avg >= 90:
        return 'A'
    elif avg >= 75:
        return 'B'
    elif avg >= 60:
        return 'C'
    else:
        return 'D'

df["Grade"] = df["Average"].apply(grade)
df

```

```python
#plot
plt.figure()
plt.bar(df["Name"],df["Average"], color = 'green')
plt.xlabel("Student")
plt.ylabel("Average marks")
plt.title("Average Marks of Students",color = 'red')
plt.show()
```

```python
df[["Maths", "Science", "English"]].mean().plot(kind = "bar", color = 'red')
plt.title("Average marks per subject")
plt.ylabel("Marks")
plt.show()
```
